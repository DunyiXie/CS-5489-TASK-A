评分使用实体起止边界与类型的精确匹配，整体报告 micro P/R/F1，另列 PER、ORG、LOC、MISC。
所有模型使用同一预测解码规则：孤立或类型不连续的 I-X 视作 B-X（与开发文档步骤 9 一致），
保存原始预测与修正次数；gold 不修正，非法 gold 直接报错。随后使用 strict IOB2 seqeval 评分，并独立核对实体计数。

训练集保持句序，5 个连续验证区块的切点放在文章边界上，避免同一文章跨训练折和验证折。
备用 3 折仅用于预先记录的 CRF 时间预算回退。Bootstrap：按原句重采样 2,000 次、95% 百分位区间；
配对比较共用采样句子，报告 B−A，不把置信区间当成 p 值。

参考：[seqeval 的模式说明](https://github.com/chakki-works/seqeval)、
[KFold](https://scikit-learn.org/1.6/modules/generated/sklearn.model_selection.KFold.html)、
[bootstrap 与配对采样](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.bootstrap.html)。
本项目将连续折切点额外对齐到文档边界；因此折大小可能不同于直接按句数等分的 KFold。

配对文件的句子编号、词和 gold 必须一一对应。
区间只反映固定模型下的句子重采样，不包含训练种子波动，也不处理同一文章内句子相关性。


```python
STEP4_COMPLETE=BOOTSTRAP_VERIFIED=BACKUP_VERIFIED=BACKUP_DOWNLOADED=False
if not PROTOCOL_VERIFIED: raise RuntimeError('Complete 4.5 first.')
TOY_ROWS=[
    {'id':'toy-0','tokens':['Jane','Villanueva','visited','London','.'],
     'tags':['B-PER','I-PER','O','B-LOC','O']},
    {'id':'toy-1','tokens':['Acme','Ltd','hired','Ana'],
     'tags':['B-ORG','I-ORG','O','B-PER']},
    {'id':'toy-2','tokens':['French','tea'],'tags':['B-MISC','O']},
    {'id':'toy-3','tokens':['No','entities','.'],'tags':['O','O','O']},
]
for row in TOY_ROWS: row['ner_tags']=[LABEL_MAPPING['label2id'][t] for t in row['tags']]
TOY_GOLD=[r['tags'] for r in TOY_ROWS]
TOY_A=[['B-PER','O','O','B-LOC','O'],['B-ORG','I-ORG','O','O'],['O','O'],['O','O','O']]
TOY_B=[['I-PER','I-PER','O','B-LOC','O'],['B-ORG','I-ORG','O','B-PER'],['B-MISC','O'],['O','O','O']]
PREDICTION_RECEIPTS={name:prediction_io.write_predictions(PREDICTION_DIR,name,'synthetic',TOY_ROWS,guess,
    LABEL_NAMES,PROTOCOL_RECEIPT['protocol_sha256']) for name,guess in [('toy_a',TOY_A),('toy_b',TOY_B)]}
CI=metrics.bootstrap_ci(TOY_GOLD,TOY_A,n_resamples=2000,confidence=0.95,seed=experiment.BOOTSTRAP_SEED)
PAIRED=prediction_io.compare_prediction_files(PREDICTION_RECEIPTS['toy_a']['path'],PREDICTION_RECEIPTS['toy_b']['path'],
    n_resamples=2000,confidence=0.95,seed=experiment.BOOTSTRAP_SEED)
IDENTICAL=metrics.paired_bootstrap(TOY_GOLD,TOY_A,TOY_A,n_resamples=2000,seed=experiment.BOOTSTRAP_SEED)
if any(v['low']!=0 or v['high']!=0 or v['estimate']!=0 for group in IDENTICAL['intervals'].values() for v in group.values()):
    raise ValueError('Identical systems must have exactly zero paired difference.')
if CI!=metrics.bootstrap_ci(TOY_GOLD,TOY_A,n_resamples=2000,confidence=0.95,seed=experiment.BOOTSTRAP_SEED):
    raise ValueError('Fixed-seed bootstrap did not reproduce.')
def interval_table(result):
    return pd.DataFrame([dict(type=kind,metric=metric,**values) for kind,group in result['intervals'].items() for metric,values in group.items()])
CI_TABLE=interval_table(CI); PAIRED_TABLE=interval_table(PAIRED)
CI_TABLE.to_csv(RESULTS_DIR/'synthetic_bootstrap_ci.csv',index=False)
PAIRED_TABLE.to_csv(RESULTS_DIR/'synthetic_paired_ci.csv',index=False)
BOOTSTRAP_CHECKS=dict(synthetic_only=True,single_system=CI,paired=PAIRED,identical_systems=IDENTICAL,
    seed_repeat_passed=True,identical_delta_zero=True,prediction_files=PREDICTION_RECEIPTS)
data_tools.write_json(RESULTS_DIR/'synthetic_bootstrap_checks.json',BOOTSTRAP_CHECKS)
BOOTSTRAP_VERIFIED=True
display(CI_TABLE)
display(PAIRED_TABLE)
print('SYNTHETIC BOOTSTRAP AND PREDICTION FILE CHECKS PASSED. No real test score was computed.')
```
